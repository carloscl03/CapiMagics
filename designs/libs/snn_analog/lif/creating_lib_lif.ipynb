```python
import pya

# 1. Load GDS layout in KLayout
layout = pya.Layout()
path_to_dir = "libs/snn_analog/lif/"
layout.read(path_to_dir + f"neurona_lvs.gds")
output_file = path_to_dir + f"neurona_lvs.lib"
cell = layout.cell("stdp")

# Define macro pins from your netlist/LEF
pins = {
    "vdd": {"dir": "inout", "type": "power"},
    "vss": {"dir": "inout", "type": "ground"},
    "Iin": {"dir": "input", "type": "signal"},
    "spike": {"dir": "output", "type": "signal"},
    "spike_neg": {"dir": "output", "type": "signal"},
}

# 2. Write minimal Liberty Stub
with open(output_file, "w") as f:
    f.write("library (neurona_lvs_lib) {\n")
    f.write("  cell (neurona_lvs) {\n")
    f.write("    area : 4874.145;\n")  # Layout size: (79.5. x 61.31)um
    
    for pin_name, info in pins.items():
        f.write(f"    pin ({pin_name}) {{\n")
        if info["type"] == "power":
            f.write("      direction : inout;\n      pg_type : primary_power;\n")
        elif info["type"] == "ground":
            f.write("      direction : inout;\n      pg_type : primary_ground;\n")
        else:
            f.write(f"      direction : {info['dir']};\n")
            f.write("      capacitance : 0.001;\n")  # Dummy stub capacitance (pF)
        f.write("    }\n")
        
    f.write("  }\n}\n")
```
