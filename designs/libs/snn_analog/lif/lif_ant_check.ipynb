```python
import os
import re
import sys
from pathlib import Path
import tempfile
import subprocess

magicrc_file = Path(os.environ['PDKPATH']) / "libs.tech" / "magic" / f"{os.environ['PDK']}.magicrc"
design_name = "neurona_lvs"
path_to_dir = "/foss/designs/libs/snn_analog/lif/"

gds_path = path_to_dir + f"{design_name}.gds"

    
magic_script_content = f"""
drc off            
gds flatglob *\\$\\$*
gds read {gds_path}

flatten {design_name}
load {design_name}
select top cell

# Se requiere la extracción previa para calcular las áreas de metal y difusión
extract do local
extract all

# Ejecución de la verificación de antena
antennacheck debug
puts "--- ANTENNA CHECK COMPLETED ---"
exit
"""

with tempfile.NamedTemporaryFile(mode='w', delete=False) as magic_script_file:
    magic_script_file.write(magic_script_content)
    magic_script_path = magic_script_file.name

# Se eliminó la coma sobrante al final de la cadena de comando
magic_cmd = f"bash -c 'magic -rcfile {magicrc_file} -noconsole -dnull < {magic_script_path}'"
magic_subproc = subprocess.run(
    magic_cmd, 
    shell=True,
    check=True,
    capture_output=True
)

magic_subproc_out = magic_subproc.stdout.decode('utf-8')
print(magic_subproc_out)


# --- PARSEO Y DETECCIÓN DE ERRORES ---

# Expresión regular para capturar líneas de violación reportadas por antennacheck
antenna_error_pattern = re.compile(r"(antenna ratio exceeded|antenna violation|failed antenna)", re.IGNORECASE)

violations = [line.strip() for line in magic_subproc_out.splitlines() if antenna_error_pattern.search(line)]

if violations:
    print(f"\n[ERROR] Se encontraron {len(violations)} violaciones de antena en '{design_name}':\n")
    for v in violations:
        print(f"  -> {v}")
    
    # Detener la ejecución del script con código de error
    sys.exit(1)
else:
    print(f"\n[SUCCESS] Succesfull antena verification for  '{design_name}'. No violations.")
```
