```python
import pya

# 1. Load GDS layout in KLayout
layout = pya.Layout()
path_to_dir = "libs/snn_analog/stdp/"
layout.read(path_to_dir + f"stdp.gds")
output_file = path_to_dir + f"stdp.lib"
cell = layout.cell("stdp")

# Define macro pins from your netlist/LEF
pins = {
    "avdd": {"dir": "inout", "type": "power"},
    "avss": {"dir": "inout", "type": "ground"},
    "nvpost": {"dir": "input", "type": "signal"},
    "nvpre": {"dir": "input", "type": "signal"},
    "vpost": {"dir": "input", "type": "signal"},
    "vpre": {"dir": "input", "type": "signal"},
    "vb_idep": {"dir": "input", "type": "signal"},
    "vb_itd": {"dir": "input", "type": "signal"},
    "vb_itp": {"dir": "input", "type": "signal"},
    "vb_pot": {"dir": "input", "type": "signal"},
    "vw": {"dir": "input", "type": "signal"},
    "Iout": {"dir": "output", "type": "signal"},
}

# 2. Write minimal Liberty Stub
with open(output_file, "w") as f:
    f.write("library (stdp_lib) {\n")
    f.write("  cell (stdp) {\n")
    f.write("    area : 8301.50;\n")  # Layout size: 107.98 x 76.88 um
    
    for pin_name, info in pins.items():
        f.write(f"    pin ({pin_name}) {{\n")
        if info["type"] == "power":
            f.write("      direction : inout;\n      pg_type : primary_power;\n")
        elif info["type"] == "ground":
            f.write("      direction : inout;\n      pg_type : primary_ground;\n")
        else:
            f.write(f"      direction : {info['dir']};\n")
            f.write("      capacitance : 0.001;\n")  # Dummy stub capacitance (pF)
        f.write("    }\n")
        
    f.write("  }\n}\n")
```
